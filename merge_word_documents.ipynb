```python
import os
from docx import Document
from docx.oxml.ns import qn
from docx.oxml import OxmlElement

def insert_page_break(doc):
    """Insert a page break in the document."""
    doc.add_page_break()

def copy_run_formatting(src_run, dest_run):
    """Copy the formatting from one run to another."""
    dest_run.bold = src_run.bold
    dest_run.italic = src_run.italic
    dest_run.underline = src_run.underline
    dest_run.font.size = src_run.font.size
    dest_run.font.name = src_run.font.name
    dest_run.font.color.rgb = src_run.font.color.rgb

def add_hyperlinked_run(dest_paragraph, src_run):
    """Add a run with a hyperlink to the destination paragraph."""
    # Create a hyperlink element
    hyperlink = OxmlElement('w:hyperlink')
    hyperlink.set(qn('r:id'), src_run._r.getparent().attrib[qn('r:id')])

    # Create a new run element
    new_run = OxmlElement('w:r')

    # Copy text
    text = OxmlElement('w:t')
    text.text = src_run.text
    new_run.append(text)

    # Add the run formatting
    for child in src_run._r:
        new_run.append(child)

    # Append the new run to the hyperlink
    hyperlink.append(new_run)

    # Append the hyperlink to the paragraph
    dest_paragraph._element.append(hyperlink)

def add_paragraph_with_formatting(src_paragraph, dest_doc):
    """Copy a paragraph from the source document to the destination document, preserving formatting and hyperlinks."""
    dest_paragraph = dest_doc.add_paragraph()

    for src_run in src_paragraph.runs:
        if "HYPERLINK" in src_run._element.xml:
            add_hyperlinked_run(dest_paragraph, src_run)
        else:
            dest_run = dest_paragraph.add_run(src_run.text)
            copy_run_formatting(src_run, dest_run)

def add_paragraphs_from_document(src_doc, dest_doc):
    """Copy paragraphs from the source document to the destination document."""
    for para in src_doc.paragraphs:
        add_paragraph_with_formatting(para, dest_doc)

def combine_word_documents(output_path, input_files):
    """Combine multiple Word documents into one."""
    combined_doc = Document()

    for index, file in enumerate(input_files):
        sub_doc = Document(file)
        
        # Add a page break before each new document, except the first one
        if index != 0:
            insert_page_break(combined_doc)
        
        # Add paragraphs from the current document, preserving formatting
        add_paragraphs_from_document(sub_doc, combined_doc)

    # Save the combined document to the specified path
    combined_doc.save(output_path)

def get_word_files_from_directory(directory):
    """Get a list of all Word files in the given directory."""
    files = [os.path.join(directory, f) for f in os.listdir(directory) if f.endswith('.docx')]
    return sorted(files)  # Sort the files alphabetically

# Specify your directory and output file
directory_path = '/Users/patescalona/Library/CloudStorage/OneDrive-Personal/Marketing Reports/CCR Reports/Centerpoint Energy/CNP Outage Tracker In Depth Interviews/Original/CNP Outage Tracker IDI Transcripts'  # Replace with your directory path
output_file = 'combined_document.docx'

# Get the list of Word files in the directory
word_files = get_word_files_from_directory(directory_path)

# Combine the documents
combine_word_documents(output_file, word_files)
```
